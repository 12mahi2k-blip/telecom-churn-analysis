# Customer Churn Analysis Walkthrough

Welcome to the interactive Jupyter Notebook for the Customer Churn Analysis project! In this notebook, we will sequentially walk through the primary data science steps:
- **Data Loading & Understanding**
- **Data Cleaning**
- **Exploratory Data Analysis (EDA)**
- **Customer Segmentation**
- **Advanced Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set stylistic choices
sns.set_theme(style="whitegrid")
```

## 1. Data Understanding

```python
df = pd.read_csv('../data/Telco_Customer_Churn.csv')

print("Shape of dataset:", df.shape)
df.head(10)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

## 2. Data Cleaning
We need to clean whitespace in columns, fix incorrect types, and handle duplicates.

```python
# Formatting column names
df.columns = [col.strip().lower().replace(' ', '_') for col in df.columns]

# Removing duplicates
df = df.drop_duplicates()

# Handling 'totalcharges' since blank spaces can exist
df['totalcharges'] = pd.to_numeric(df['totalcharges'], errors='coerce')
df['totalcharges'] = df['totalcharges'].fillna(0)

print("Data cleaning successful.")
```

## 3. Exploratory Data Analysis (EDA)

```python
df.describe()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='churn', palette='Set2')
plt.title('Customer Churn Distribution')
plt.show()
```

## 4. Customer Segmentation

```python
def get_tenure_group(tenure):
    if tenure <= 12:
        return '0-12 Months'
    elif tenure <= 36:
        return '13-36 Months'
    else:
        return '37+ Months'

df['tenure_group'] = df['tenure'].apply(get_tenure_group)
group_counts = df['tenure_group'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(group_counts, labels=group_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette("pastel"))
plt.title('Segmentation by Tenure')
plt.show()
```

## 5. Advanced Analysis

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='contract', hue='churn', palette='magma')
plt.title('Churn by Contract Type')
plt.show()
```
